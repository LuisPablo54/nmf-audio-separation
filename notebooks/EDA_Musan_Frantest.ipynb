```python
import os
import librosa
import librosa.display
import numpy as np
import matplotlib.pyplot as plt

# 1. Tus directorios base

musan_path   = r'C:\Users\frank\OneDrive\Documentos\Ibero Semestres\6tosemestre\Python\data\musan'
output_path  = r'C:\Users\frank\OneDrive\Documentos\Ibero Semestres\6tosemestre\Python\OptimizacionM\nmf-audio-separation\outputs'


# 2. Construir las rutas a los archivos WAV
# Nota: Ajusta 'us-gov/speech-0001.wav' al archivo exacto que quieras usar de tu carpeta
ruta_speech = os.path.join(musan_path, 'speech', 'us-gov', 'speech-0001.wav')
ruta_noise = os.path.join(musan_path, 'noise', 'free-sound', 'noise-0001.wav')

# 3. Cargar los audios (forzamos 22050 Hz para que ambos coincidan)
y_speech, sr = librosa.load(ruta_speech, sr=22050)
y_noise, _ = librosa.load(ruta_noise, sr=22050)

# 4. Igualar tamaños (recortar al tamaño del más corto) para poder sumarlos
min_len = min(len(y_speech), len(y_noise))
y_speech = y_speech[:min_len]
y_noise = y_noise[:min_len]

# 5. Mezclar digitalmente
y_mix = y_speech + y_noise

# 6. Calcular el STFT para la mezcla
D_mix = librosa.stft(y_mix, n_fft=1024, hop_length=256)

# 7. Obtener la magnitud (Matriz X)
X_mix = np.abs(D_mix)

print(f"Dimensiones de la matriz X (F x T): {X_mix.shape}")
print(f"Valor mínimo de X: {X_mix.min()} (debe ser >= 0 para la NMF)")

# 8. Visualización
X_mix_db = librosa.amplitude_to_db(X_mix, ref=np.max)

plt.figure(figsize=(10, 4))
librosa.display.specshow(X_mix_db, sr=sr, hop_length=256, x_axis='time', y_axis='hz')
plt.colorbar(format='%+2.0f dB')
plt.title('Espectrograma de la Mezcla (Voz + Ruido)')
plt.tight_layout()

# Opcional: Guardar la gráfica en tu carpeta de outputs
# plt.savefig(os.path.join(output_path, 'eda_espectrograma_mezcla.png'))

plt.show()
```
